## a) Compare Results on MNIST and Fashion MNIST

MNIST reached a final validation accuracy of 0.9699 against 0.8712 for Fashion-MNIST. MNIST results were more accuracy by about 10%, showing that this dataset was easier for the network to classify. This may be because handwritten digits are simple and high-contrast, and each digit has a distinct and relatively consistent shape. Fashion-MNIST images have similar shapes and sihouttes even when they belong to different classes, making boundaries harder to learn. 

## b) Compare Model Configurations

Doubling the hidden units raised validation accuracy on both datasets: +0.15 points on MNIST, and +1.48 points on Fashion-MNIST. MNIST already performs really well with 64 units, so the extra units make less difference. Fashion-MNIST benefis more from extra neurons. The 128-unit models also reached higher training accuracy.

The effect of switching from ReLU to tanh depended on the dataset. On MNIST, tanh was slightly worse (-0.58 points). On Fashion-MNIST, tanh was slightly better (+0.85 points).

Short runs don't prove one configuration is always better. Training isn't finishes as the models haven't converged. So the results only compare how fast each model learns, not its final performance. A configuration that learns more slowly could still perform better after more epochs.

## c) Dense Versus Conv1D Layers

In a Dense Layer, every input connects to every neuron, and each connection has its own weight. A dense neuron leans one weight per feature, then sums them all at once. The layer treats its input as an unordered collection of numbers.

A Conv1D layer looks only at a small window of neighboring inputs. The same filter (small set of weights) slides along the input and is reused at evry position. A filter that learns a pattern will find that pattern anywhere in the sequence. It requires far fewer parameters than a dense layer.

Local structure helps with images and sequences because neighboring data are related. Additionally, patterns are local and can appear anywhere in the data. In an image, local patterns include things like edges. This means Conv1D layer can learn faster with fewer positions.

The Iris Dataset doesn't have this structure because the features are unrelated. The relative position of the columns doesn't actually mean anything. 

## d) Overfitting versus Underfitting

Iris Conv1D model from Question 2: The training loss and validation loss are both still clearly dropping at epoch 10 with no sign of levelling off. The two curves stay close together the whole time. The model continues to grow more confident in predictions it already has right. This model demonstrates signs of underfitting. 

One change to investifate: train for many more epochs, with early stopping. This is the simplest and easiest fix and does not require changing the architecture. If longer training starts to cause overfitting, early stopping will prevent it from continuing. This will most likely improve the accuracy.
