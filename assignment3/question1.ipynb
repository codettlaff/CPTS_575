# Part 1: NBA Data
## Read the Data File

```python
import os
import pandas as pd

base_dir = os.getcwd()
data_dir = os.path.join(base_dir, 'data')
nba_data_filepath = os.path.join(data_dir, 'NBA_Stats_23_24.csv')

# Load Data.
nba = pd.read_csv(nba_data_filepath, encoding='latin1')

# Display the first few rows and column names.
display(nba.head())
print(nba.columns.tolist())
```

## Part a. Check for Missing Values

```python
# 1.a Check for missing values
print(f"Missing values:\n{nba.isna().sum()}\n")
```

## Part b. Sort Players
### Count players with Free Throws > 0.5 and Assists Per Game > 0.7

```python
# 1.b Count players with FT > 0.5 and AST > 0.7 
qualified_players = nba.query("FT > 0.5 and AST > 0.7")
print(f"Number of players with FT > 0.5 and AST > 0.7: {len(qualified_players)}")
```

### Rank players by points

```python
# Top 10 players by points
top_10_points = (
    nba.loc[:, ['Player', 'PTS', 'Tm', 'FG', 'TOV']]
    .sort_values('PTS', ascending=False)
    .head(10))

print(f"Top 10 players by points:\n{top_10_points}\n")
print(f"Player with the seventh highest points: {top_10_points.iloc[6]['Player']}")
```

## Part c. Add new columns to the dataframe

```python
# 1.c Add FGP and FTP
nba = nba.assign(
    FGP=(nba['FG'] / nba['FGA'] * 100).round(2),
    FTP=(nba['FT'] / nba['FTA'] * 100).round(2))

# Display new dataframe
display(nba[['Player', 'FG', 'FGA', 'FGP', 'FT', 'FTA', 'FTP']].head())
```

### Find FGP and FTP for Josh Giddey

```python
# FGP and FTP for Josh Giddey
josh_giddey = nba.loc[
    nba['Player'] == 'Josh Giddey',
    ['Player', 'FGP', 'FTP']]

print(f"Josh Giddey FGP and FTP:\n {josh_giddey}")
```

## Part d. Find average, min, and max Offensive Rebounds per Game for each team.

```python
# 1.d Offensive rebounds by team
nba = nba.assign(
    FGP=(nba['FG'] / nba['FGA'] * 100).round(2),
    FTP=(nba['FT'] / nba['FTA'] * 100).round(2))

team_orb = (
    nba.groupby('Tm')['ORB']
    .agg(['mean', 'min', 'max'])
    .sort_values('mean', ascending=False))

print(f"Offensive rebounds per game by team:\n{team_orb}\n")
print(f"Team with the maximum Offensive rebounds per game: {team_orb['max'].idxmax()}\n")
```

## Part e. Impute the missing FTP values
### Impute missing values as FGP multiplied by average FTP for that team.
This method assumes a player's free-throw percentage is proportional to their field-goal average, while also accounting for differences between teams. This method relies on both FGP and the team's average FTP. Eight players with missing FTP values also have missing FGP. This results in eight remaining missing values.

```python
# 1.e Impute missing FTP using FGP and team average FTP
team_average_ftp = nba.groupby('Tm')['FTP'].transform('mean')
nba_imputed_fgp = nba.assign(
    FTP=nba['FTP'].fillna(
       nba['FGP'] / 100 * team_average_ftp))

print(f"Number of missing FTP values after imputation:", nba_imputed_fgp['FTP'].isna().sum())
```

### Impute missing values as average FTP for that team.
This method assumes a player with a missing FTP would have a free-throw percentage similar to an average player on their team. This method only uses the team's average FTP, so it can fill every missing FTP value, with no remaining missing values.

```python
# 1.f Impute missing FTP using team average FTP only
nba_imputed_mean = nba.assign(
    FTP=nba['FTP'].fillna(
        nba.groupby('Tm')['FTP'].transform('mean')))

print(f"Number of missing FTP values after imputation:", nba_imputed_mean['FTP'].isna().sum())
```

### Compare Methods
The first method incorporates more information, potentially providing more accurate imputed values. However, it is more complex, resulting in some remaining missing values. The second method is more simple, filling all missing values, but the imputed values may be less accurate. A potentially better method would incorporate other variables such as FTA, FT, or FGP to better account for differences between individual players. However, additional logic would need to be incorporated to make sure all missing values are filled, even if these other variables also have missing values.

```python
# Compare the two imputation methods specifically on rows that originally had missing FTP.
comparison = nba.loc[nba['FTP'].isna(), ['Player', 'Tm', 'FGP', 'FTP']].copy()
comparison['FTP_FGP_method'] = nba_imputed_fgp.loc[comparison.index, 'FTP']
comparison['FTP_team_mean_method'] = nba_imputed_mean.loc[comparison.index, 'FTP']

print(f"Comparison of imputation methods:\n{comparison}")
```
